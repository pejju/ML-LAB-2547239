## Task 4: AQI Distribution Analysis

In this section, we examine the distribution of Air Quality Index (AQI) values across India. Understanding the distribution helps determine if poor air quality is a systemic issue across most cities or driven by a few extreme outliers.

```python
import scipy.stats as stats

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.histplot(aqi['AQI'], bins=50, kde=True, ax=axes[0], color='skyblue')

axes[0].set_title('Distribution of AQI across Indian Cities')
axes[0].set_xlabel('AQI Value')
axes[0].set_ylabel('Frequency')

sns.boxplot(x=aqi['AQI'], ax=axes[1], color='lightcoral')

axes[1].set_title('Spread and Outliers of AQI')
axes[1].set_xlabel('AQI Value')

plt.tight_layout()
plt.show()

mean_aqi = aqi['AQI'].mean()
median_aqi = aqi['AQI'].median()
skewness = aqi['AQI'].skew()

print(f"Mean AQI: {mean_aqi:.2f}")
print(f"Median AQI: {median_aqi:.2f}")
print(f"Skewness: {skewness:.2f}")
```

## Task 5: Outlier Detection And Treatment

Outliers in AQI data often represent extreme pollution events (like crop burning or festivals). While these are real data points, they can disproportionately influence statistical models. Here we detect and treat them to ensure robust analysis.

```python
Q1 = aqi['AQI'].quantile(0.25)
Q3 = aqi['AQI'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = aqi[(aqi['AQI'] < lower_bound) | (aqi['AQI'] > upper_bound)]

print("Number of outliers detected:", len(outliers))
print(f"Lower Bound for AQI: {lower_bound:.2f}")
print(f"Upper Bound for AQI: {upper_bound:.2f}")

# Outlier Treatment using Winsorization (Capping)

aqi_capped = aqi.copy()

aqi_capped['AQI'] = np.where(
    aqi_capped['AQI'] > upper_bound,
    upper_bound,
    aqi_capped['AQI']
)

aqi_capped['AQI'] = np.where(
    aqi_capped['AQI'] < lower_bound,
    lower_bound,
    aqi_capped['AQI']
)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(x=aqi['AQI'], ax=axes[0], color='lightcoral')
axes[0].set_title('AQI Before Treatment')

sns.boxplot(x=aqi_capped['AQI'], ax=axes[1], color='lightgreen')
axes[1].set_title('AQI After Treatment')

plt.tight_layout()
plt.show()

aqi = aqi_capped
```

# Outlier Analysis and Treatment Rationale

**Method Selected:** Interquartile Range (IQR).

**Justification:** IQR is a robust method for detecting outliers in non-normally distributed data.

**Treatment Strategy:** Capping (at the 1.5 * IQR boundary).

**Rationale:** We chose capping over removal because extreme pollution events are geographically and chronologically significant. Deleting them would hide the severity of India's pollution peaks. Capping reduces their statistical 'pull' while keeping the record of those days in our dataset.

**Results:** The 'After Treatment' boxplot shows a more controlled distribution, which will provide more reliable correlation results when we merge this with the agricultural data.

## Task 6: AQI Trend Over Time

Is air quality improving or worsening? In this section, we analyze the annual average AQI from 2015 onwards to identify long-term environmental trends in India.

```python
aqi['Date'] = pd.to_datetime(aqi['Date'])

aqi['Year'] = aqi['Date'].dt.year

yearly_aqi = aqi.groupby('Year')['AQI'].mean().reset_index()

print(yearly_aqi)

plt.figure(figsize=(10, 5))

sns.lineplot(data=yearly_aqi, x='Year', y='AQI', marker='o')

plt.title('Average Annual AQI Trend in India')
plt.xlabel('Year')
plt.ylabel('Average AQI')

plt.grid(True)

plt.show()

overall_change = yearly_aqi['AQI'].iloc[-1] - yearly_aqi['AQI'].iloc[0]

print(f"Overall AQI Change: {overall_change:.2f}")

if overall_change < 0:
    print("Inference: Air quality has improved over the years.")
else:
    print("Inference: Air quality has worsened over the years.")
```

### Report: India's Air Quality Trend (2015–2020)

Our analysis of AQI data collected from major Indian cities highlights several important environmental trends over the years.

- **Highest Pollution Period:** The data indicates that air quality was poorest during **2015–2016**, when the average AQI values were at their maximum. This suggests that pollution levels were considerably higher during these years.

- **Gradual Improvement:** After 2016, the overall AQI trend shows a steady decline, indicating an improvement in air quality. Notably, **2020 recorded the lowest average AQI**, reflecting a significant reduction in pollution levels.

- **Year-to-Year Variations:** Although the general trend points toward cleaner air, the improvement was not uniform across all years. Minor fluctuations can be observed, demonstrating that environmental conditions and human activities continue to influence air quality.

- **Overall Observation:** The long-term pattern suggests progress in reducing pollution levels. However, AQI values remain high enough in several regions to require sustained environmental monitoring and pollution-control measures.

**Conclusion:** Compared to the beginning of the study period, India's air quality has shown noticeable improvement. Nevertheless, continued efforts are necessary to achieve consistently healthy air quality standards across the country.

## Task 7: Seasonal AQI Analysis

This section focuses on analyzing seasonal variations in air quality across India. By comparing average AQI levels across different seasons, we aim to determine whether the Harvest/Post-Monsoon season (October–December) experiences significantly higher pollution levels than the rest of the year.

```python
aqi['Month'] = aqi['Date'].dt.month

def get_season(month):
    if month in [3, 4, 5]:
        return 'Summer'
    elif month in [6, 7, 8, 9]:
        return 'Monsoon'
    elif month in [10, 11, 12]:
        return 'Harvest/Post-Monsoon'
    else:
        return 'Winter'

aqi['Season'] = aqi['Month'].apply(get_season)

seasonal_aqi = aqi.groupby('Season')['AQI'].mean().reset_index()

print(seasonal_aqi)

plt.figure(figsize=(8, 5))

sns.barplot(data=seasonal_aqi, x='Season', y='AQI')

plt.title('Average AQI by Season')
plt.xlabel('Season')
plt.ylabel('Average AQI')

plt.xticks(rotation=15)

plt.show()

highest_pollution_season = seasonal_aqi.loc[seasonal_aqi['AQI'].idxmax(), 'Season']

print(f"Season with Highest AQI: {highest_pollution_season}")
```

### Evaluation of the NGO's Claim

**Claim:** *"Air pollution reaches its highest levels during the October–December harvest season."*

**Assessment:** The claim is supported by the observed data.

#### Key Findings

- The **Harvest/Post-Monsoon season (October–December)** recorded the highest average AQI among all seasons analyzed.
- Elevated pollution levels during this period may be associated with factors such as crop residue burning, increased particulate matter, and weather conditions that limit pollutant dispersion.
- The **Winter season** also exhibited relatively high AQI values, likely due to lower temperatures and atmospheric inversion effects that trap pollutants near the surface.
- In contrast, the **Monsoon season** showed the lowest AQI levels, as rainfall helps remove pollutants from the atmosphere.

#### Conclusion

The seasonal analysis indicates that air quality deteriorates significantly during the Harvest/Post-Monsoon period. Therefore, the available evidence supports the NGO's claim that the October–December season experiences the highest pollution levels in India.

## Task 8: Dataset Merging and Relationship Analysis

### Challenges in Direct Dataset Integration

The AQI and crop production datasets cannot be merged directly because they are recorded at different levels of detail.

1. **AQI Dataset:** Contains observations collected daily for individual cities.
2. **Crop Dataset:** Contains agricultural statistics reported annually (or seasonally) at the state and district levels.

Since the datasets differ in both temporal and geographical granularity, a direct comparison would not produce meaningful results.

To establish a common basis for analysis, the AQI data is aggregated to obtain the **average AQI for each state**. This transformation aligns the air quality data with the crop production dataset, enabling further relationship and correlation analysis.

```python
state_aqi_avg = aqi.groupby('State_Name')['AQI'].mean().reset_index()

state_crop_avg = crop.groupby('State_Name')[['Area', 'Production']].mean().reset_index()

merged_df = pd.merge(state_crop_avg, state_aqi_avg, on='State_Name', how='inner')

print(f"Merged Dataset Shape (by State): {merged_df.shape}")
print("Common States analyzed:", merged_df['State_Name'].unique())

corr_matrix = merged_df[['Area', 'Production', 'AQI']].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)

plt.title('Correlation Heatmap: Average State Agriculture vs. Average AQI')

plt.show()
```

# Relationship Analysis Insights

The correlation analysis provides useful observations regarding the association between air quality and agricultural indicators.

1. **AQI vs. Production:** The correlation between AQI and crop production helps evaluate whether changes in air quality are associated with agricultural output. A negative correlation would suggest that higher pollution levels may coincide with lower production levels.

2. **Area vs. Production:** A positive relationship is generally expected between cultivated area and crop production, as larger agricultural areas typically contribute to greater overall output.

**Important Note:** Correlation only measures the strength and direction of an association between variables. It does not establish a cause-and-effect relationship. Factors such as rainfall, irrigation facilities, soil fertility, farming practices, and technological advancements may also significantly influence agricultural productivity.
